```python
# Import Colab userdata
from google.colab import userdata
# Step2: Retrieve Gemini API Key
api_key = userdata.get("Gemini_API_Key")
#Verify API Key
if not api_key:
  raise ValueError("Gemini API Key not found!")

print("Gemini API Key loaded sucessfully.")
```

```python
# Import Gemini SDK
from google import genai
#Intiallize Gemini client to establish the connection with gemini Ai server
client = genai.Client(api_key=api_key)
print("Gemini client intiallized successfully.")
```

```python
#step 4: Send prompt to Gemini model
response = client.models.generate_content(
    model = "gemini-3.5-flash",
    contents = "What is Artificial intilligence? Explain in simple words."

)
```

```python
#step 5: Display generated response
print("\nLLM Response:")
print(response.text)
```
